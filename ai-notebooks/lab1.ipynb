**Задание 1.**
Создать 8x8 матрицу и заполнить её в шахматном порядке нулями и 
единицами. 

```python
import numpy as np
```

```python
rows, cols = np.indices((8, 8))

print(f'rows\n{rows}\n')
print(f'cols\n{cols}\n')

matrix = rows + cols
print(f'matrix\n{matrix}\n')


chess_matrix = matrix % 2
print(f'chess_matrix\n{chess_matrix}')
```

#  Работа с числовыми данными

Загрузить фрейм данных по ссылке: 
https://raw.githubusercontent.com/akmand/datasets/master/iris.csv. 
Необходимо выполнить нормализацию первого числового признака 
(sepal_length_cm) с использованием минимаксного преобразования, а 
второго (sepal_width_cm) с задействованием z-масштабирования. 

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler
```

```python
url = "https://raw.githubusercontent.com/akmand/datasets/master/iris.csv"
df = pd.read_csv(url)
```

```python
print("Первые 5 строк исходных данных:")
print(df.head())
print("\nОписательная статистика до нормализации:")
print(df[['sepal_length_cm', 'sepal_width_cm']].describe())
```

```python
min_max_scaler = MinMaxScaler()
df['sepal_length_normalized_minmax'] = min_max_scaler.fit_transform(df[['sepal_length_cm']])
```

```python
standard_scaler = StandardScaler()
df['sepal_width_standardized_z'] = standard_scaler.fit_transform(df[['sepal_width_cm']])
```

```python
print("\nРезультаты после нормализации (первые 5 строк):")
print(df[['sepal_length_cm', 'sepal_length_normalized_minmax', 
          'sepal_width_cm', 'sepal_width_standardized_z']].head())
```
